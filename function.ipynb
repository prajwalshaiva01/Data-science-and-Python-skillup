```python
## variable length argument
## positional and keyword arguments
def print_numbers(*prajwal):
    for numbers in prajwal:
        print(numbers)
```

```python
print_numbers(1,2,3,'prajwal')
```

```python
def print_numbers(*args):
    for numbers in args:
        print(numbers)
```

```python
print_numbers(1,2,3,'prajwal')
```

```python
## keyword argu
def print_detalis(**kwargs):
    for key,value in kwargs.items():
        print(f'{key}:{value}')
```

```python
print_detalis(name='prajwal',age='23')
```

```python
def print_detalis(*args,**kwargs):
    for val in args:
        print(f' positional arument {val}')

    for key,value in kwargs.items():
            print(f'{key}:{value}')
```

```python
print_detalis(1,2,3,'prajwal',name='prajwal',age='23',country='india')
```
