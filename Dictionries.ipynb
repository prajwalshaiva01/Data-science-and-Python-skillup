```python
## nested dict
student={
    'student1':{'name':'prajwal','age':23},
    'student2':{'name':'kiran','age':22},
}
print(student)
```

```python
## acces nested dict 
print(student['student2'],['name'])
print(student['student2'],['age'])
```

```python
## iterating over dict
for student_id,student_info in student.items():
    print(f'(student_id):{student_info}')
    for keys,values in student_info.items():
        print(f'(keys):{values}')
```

```python
numbers=[1,2,2,3,3,3,4,4,4,4,4]
frequency={}
for numbers in numbers:
    if numbers in frequency:
        frequency[numbers]+=1
    else:
        frequency[numbers]=1
print(frequency)
```
