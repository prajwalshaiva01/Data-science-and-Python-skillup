```python
## while loop
count=0
while count<8:
    print(count)
    count=count+1
```

```python
count=0
while count%2==0:
    print(count)
    count=count+1
```

```python
## loop control ststment
##break
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue
for i in range (10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass 
for i in range(10):
    if i==3:
        pass
    print(i)
```

```python
## nested loop
for i in range(3):
    for j in range(2):
        print(f'i:{i} and j:{j}')
```
